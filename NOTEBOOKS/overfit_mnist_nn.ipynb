```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense,Flatten
```

```python
(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data()
```

```python
X_train = X_train/255
X_test = X_test/255
```

```python
X_train[0]
```

```python
model = Sequential()

model.add(Flatten(input_shape=(28,28))) # as we want all the the 28*28 shape to chage to 784*1, that is why we use the flatten layer to convert higher dimentional to 1 dimentional
model.add(Dense(128, activation='relu'))
model.add(Dense(32, activation='relu'))
model.add(Dense(10, activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(loss = 'sparse_categorical_crossentropy', optimizer= 'Adam', metrics=['Accuracy']) # we could've used categorical crossentropy pr usme labels ko one hot encpde krna padta hai isme no need
```

```python
history = model.fit(X_train , y_train, epochs = 10, validation_split=0.2)
```

```python
y_prob = model.predict(X_test)
y_pred = y_prob.argmax(axis=1)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_pred)
```

```python
import matplotlib.pyplot as plt
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
```
