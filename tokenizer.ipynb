```python
from collections import defaultdict

lst = tokens

vocab = set(lst)
counter = 0
max_val = 255 # in utf-8 this is the max value
found = True

while found:
    m = defaultdict(int)
    i = 0

    while i < len(lst) - 1:
        pair = (lst[i], lst[i + 1])
        m[pair] += 1
        i += 1

    v = sorted(m.items(), key=lambda item: item[1], reverse=True)

    if not v or v[0][1] < 2:
        found = False
    else:
        target_pair = v[0][0]
        new_token = max_val + 1 + counter
        vocab.add(new_token)
        counter += 1

        b = []
        a = 0
        while a < len(lst):
            if a < len(lst) - 1 and (lst[a], lst[a + 1]) == target_pair:
                b.append(new_token)
                a += 2
            else:
                b.append(lst[a])
                a += 1

        lst = b
print("Final List:", lst)
print("Vocabulary:", vocab)
print(len(lst))
```

```python
# how to decode the tokens ??
# in the necoder block aboe we will now create a merges dictionary to keep track of what was encoded to what
from collections import defaultdict
counter = 0
lst = tokens.copy()
vocab = set(lst)
max_val = 255  # Base UTF-8 byte range (0-255)

# Dictionary to store token mappings: new_token_id -> (pair_left, pair_right)
merges = {}

while max(vocab) < 277:
    m = defaultdict(int)
    i = 0
    while i < len(lst) - 1:
        pair = (lst[i], lst[i + 1])
        m[pair] += 1
        i += 1

    v = sorted(m.items(), key=lambda item: item[1], reverse=True)

    if not v or v[0][1] < 2:
        break

    target_pair = v[0][0]
    new_token = max_val + 1 + counter
    counter += 1
    vocab.add(new_token)

    # SAVE THE MERGE PAIR HERE
    merges[new_token] = target_pair

    b = []
    a = 0
    while a < len(lst):
        if a < len(lst) - 1 and (lst[a], lst[a + 1]) == target_pair:
            b.append(new_token)
            a += 2
        else:
            b.append(lst[a])
            a += 1
    lst = b
```

```python
def decode_token(token_id):
    # Base case: if it's a raw UTF-8 byte (0-255), return it directly
    if token_id <= 255:
        return [token_id]

    # Recursive case: split the token back into its parent pair and expand both
    left, right = merges[token_id]
    return decode_token(left) + decode_token(right)

def decode(token_list):
    # 1. Expand all tokens into raw byte values (0-255)
    raw_bytes = []
    for token in token_list:
        raw_bytes.extend(decode_token(token))

    # 2. Convert byte integers back to a UTF-8 string
    return bytes(raw_bytes).decode("utf-8")
```

```python
# we have the merges dictionary, and we need to follow the order of those merges
def encode(text):
    tokens = list(text.encode("utf-8"))

    # 2. Apply merges in chronological training order
    for new_token, pair in merges.items():
        i = 0
        new_tokens = []
        while i < len(tokens):
            # Check if adjacent pair matches current merge rule
            if i < len(tokens) - 1 and (tokens[i], tokens[i + 1]) == pair:
                new_tokens.append(new_token)
                i += 2  # Skip both merged elements
            else:
                new_tokens.append(tokens[i])
                i += 1
        tokens = new_tokens

    return tokens
```

```python
# Part 1: The Core Differences (tiktoken vs. SentencePiece)
# The decision to use one over the other comes down to two main factors: what you are trying to do, and how the algorithm fundamentally reads the text.

# 1. Training vs. Inference (Read-Only vs. Read & Write)

# tiktoken (OpenAI): This is a "read-only" tool built for inference. If you want to use an existing model like GPT-4, you use tiktoken to apply OpenAI's pre-existing vocabulary to your text. You cannot use it to train a new vocabulary from scratch.

# SentencePiece (Google): This is a "read and write" tool built for training and inference. If you are building a brand-new LLM from scratch, SentencePiece has the algorithms required to scan your training data and build a completely custom vocabulary.

# 2. The Order of Operations (Bytes vs. Code Points)

# tiktoken (Bytes First): It takes your text and immediately smashes it down into raw UTF-8 bytes (1s and 0s). It then looks for patterns and merges those bytes together to form tokens.

# SentencePiece (Code Points First): It looks at the text as Unicode code points (the unique character ID numbers, like U+0068 for 'h'). It merges these conceptual characters together to form tokens, only resorting to bytes in an emergency.

# Part 2: SentencePiece's Superpowers (The Settings)
# Because SentencePiece is used to train new models, you have to configure how it handles the quirks of human language. Here is what those specific settings in the code actually do:

# 1. Character Coverage (character_coverage)
# Think of this as the VIP list for your starting vocabulary. When SentencePiece scans your training data, it counts every unique character.

# If you set this to 1.0 (100%), every single weird typo or rare symbol gets a slot in your vocabulary, which wastes space.

# By setting it to 0.9995 (like Andrej does), you tell the tokenizer to include the most frequent characters that make up 99.95% of your text, and completely ignore the incredibly rare 0.05%.

# 2. Byte Fallback (byte_fallback)
# Think of this as the model's emergency backup plan for that missing 0.05%.

# Normally, if a model sees a character it doesn't recognize (like a rare emoji or foreign symbol), it just outputs a useless "Unknown" token.

# If you set byte_fallback=True, the tokenizer says, "I don't recognize this code point, so I will chop it down into its underlying raw UTF-8 bytes and process those instead!" This ensures your model can technically process any text input without losing data.

# 3. Special Tokens
# Special tokens are invisible instructions that tell the AI about the structure of the text. The standard ones are <unk> (Unknown), <bos> (Beginning of Sequence), <eos> (End of Sequence), and <pad> (Padding, used to fill empty space when training multiple sentences at once).

# How to turn them ON or OFF: In SentencePiece, you don't use "True/False" to toggle these. You assign them an ID number.

# Giving a token a positive number (like bos_id=1) turns it ON.

# Setting a token's ID to -1 turns it OFF.

# (Note: The <unk> token is strictly required by SentencePiece to exist, so it is usually set to unk_id=0 and can never be turned off!)
```

```python
# train a sentencepiece model on it
# the settings here are (best effort) those used for training Llama 2
import os

options = dict(
  # input spec
  input="toy.txt",
  input_format="text",
  # output spec
  model_prefix="tok400", # output filename prefix
  # algorithm spec
  # BPE alg
  model_type="bpe",
  vocab_size=400,
  # normalization
  normalization_rule_name="identity", # ew, turn off normalization
  remove_extra_whitespaces=False,
  input_sentence_size=200000000, # max number of training sentences
  max_sentence_length=4192, # max number of bytes per sentence
  seed_sentencepiece_size=1000000,
  shuffle_input_sentence=True,
  # rare word treatment
  character_coverage=0.99995,
  byte_fallback=True,
  # merge rules
  split_digits=True,
  split_by_unicode_script=True,
  split_by_whitespace=True,
  split_by_number=True,
  max_sentencepiece_length=16,
  add_dummy_prefix=True,
  allow_whitespace_only_pieces=True,
  # special tokens
  unk_id=0, # the UNK token MUST exist
  bos_id=1, # the others are optional, set to -1 to turn off
  eos_id=2,
  pad_id=-1,
  # systems
  num_threads=os.cpu_count(), # use ~all system resources
)

spm.SentencePieceTrainer.train(**options)
```
